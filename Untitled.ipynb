```python
from automata.fa.dfa import DFA

# d : automate des mots sur {a,b} dont le nombre de a est multiple de 3
d = DFA(
    states={"0", "1", "2"}, input_symbols={"a", "b"},
    transitions={
    "0": {"a": "1", "b": "0"},
        "1": {"a": "2", "b": "1"},
        "2": {"a": "0", "b": "2"},
    },
    initial_state="0", final_states={"0"},
)
```

```python
from automata.fa.nfa import NFA

# n : NFA avec une epsilon-transition
n = NFA(
states={"p", "q", "r"}, input_symbols={"a", "b"},
transitions={
"p": {"a": {"p", "q"}, "": {"r"}},
"q": {"b": {"q"}},
"r": {"a": {"r"}},
},
initial_state="p", final_states={"q", "r"},
)
```

**accepts_input**
Teste si un mot est accepté (True/False)

```python
from automata.fa.dfa import DFA
d.accepts_input("aaa")
# -> True
```

**read_input**
Comme accepts_input, mais lève une exception si le mot est rejeté.

```python
from automata.fa.dfa import DFA
d.read_input("aaa")
# -> '0'
```

**read_input_stepwise**  
Génère la suite des états traversés, un pas à la fois.

```python
from automata.fa.dfa import DFA
list(d.read_input_stepwise("aaa"))
# -> ['0', '1', '2', '0']
```

**copy**
Copie profonde de l’automate. Fonctionne aussi correctement sur ExtendedDFA/ExtendedNFA depuis
la correction de **input_parameters**

```python
from automata.fa.dfa import DFA
d2 = d.copy()
d2 == d
# -> True
```

**validate**

**__contains__ (mot in automate)**  
Sucre syntaxique pour accepts_input.

```python
from automata.fa.dfa import DFA
"aaa" in d
# -> True
```

**iter_transitions**  
Itère sur les transitions sous forme de triplets (état, état, symbole).

```python
from automata.fa.dfa import DFA
sorted(d.iter_transitions())[:2]
# -> [('0', '0', 'b'), ('0', '1', 'a')]
```

**show_diagram**  
Construit un diagramme graphviz de l’automate (nécessite pygraphviz).

```python
d.show_diagram()
```

**is_minimal**  
Teste si l’automate est déjà minimal (accessible + toutes les classes sont des singletons).

```python
d.is_minimal()
# -> True
```
